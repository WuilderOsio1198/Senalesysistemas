```python
%%writefile THD.py
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal as sig
import sympy as sym
st.sidebar.title("Presentado por:")
st.sidebar.write("Wuilder Joafran Osio Tandioy")


# Título y descripción general
st.title("Análisis de un péndulo elástico amortiguado")

st.write("""
Las propiedades mecánicas del péndulo elástico amortiguado se analizan utilizando la teoría de los SLIT. La configuración mecánica subyacente se muestra en la siguiente figura:


![Péndulo Elástico Amortiguado](https://github.com/amalvarezme/SenalesSistemas/blob/master/4_Transformada_Laplace/damped_spring.png?raw=1)
""")


# Sección teórica
st.write("""
### Ecuación diferencial

La ecuación diferencia del sistema mecánico se obtiene al considerar el equilibrio de fuerzas ejercidas sobre la masa:
""")

st.latex(r"F_\text{S}(t) + F_\text{F}(t) + F_\text{I}(t) = F_\text{E}(t)")

st.write("""
Donde:

Donde $F_{E}(t) $ denota una fuerza externa que actúa sobre la masa. La fuerza $F_{S}(t)$ inducida por el resorte está dada por la ley de Hooke
""")
st.latex(r"F_\text{S}(t) = k y(t)")

st.write("""

Es común modelar la fuerza de fricción $F_{F}(t)$ inducida por el amortiguador de manera proporcional a la velocidad de la masa
""")
st.latex(r" F_\text{F}(t) = c \frac{d y(t)}{dt}")

st.write("""
La fuerza inercial $F_{I}(t)$ debida a la aceleración de la masa se da como
""")
st.latex(r" F_\text{I}(t) = m \frac{d^2 y(t)}{dt^2} ")

st.write("""
Al introducir las fuerzas anteriormente halladas en el equilibrio de fuerzas se obtiene la ecuación diferencial que describe el desplazamiento del péndulo elástico amortiguado
""")

st.latex(r"m\,\frac{d^{2}y(t)}{dt^{2}} + c\,\frac{dy(t)}{dt} + k\,y(t) = F_{E}(t)")


st.write("""
Como consecuencia de la fuerza externa.

La ecuación anterior constituye una EDO con coeficientes constantes. Puede interpretarse como un SLTI con la fuerza externa $x(t)= F_{E}(t)$ como señal de entrada y el desplazamiento de la masa como la señal de salida $y(t)$.
""")

# Inicializar impresión simbólica
sym.init_printing()

# Definir símbolos
t, m, c, k = sym.symbols('t m c k', real=True)
x = sym.Function('x')(t)
y = sym.Function('y')(t)

# Crear la ecuación diferencial
ode = sym.Eq(m * y.diff(t, 2) + c * y.diff(t) + k * y, x)

# Mostrar en pantalla
st.write("### Ecuación diferencial generada con SymPy:")
st.latex(sym.latex(ode))

st.write("""
Tomando transformada de Laplace (condiciones iniciales cero) y usando
$Y(s)$ Y $F(s)$ donde: """)


st.latex(r" H(s) = \frac{Y(s)}{F(s)}")

st.write("Función de transferencia")

st.latex(r"H(s)=\frac{1}{m s^2 + c s + k}")


# Título y descripción general
st.title("Análisis de un péndulo elástico amortiguado")

st.write("""
Las propiedades mecánicas del péndulo elástico amortiguado se analizan utilizando la teoría de los SLIT. La configuración mecánica subyacente se muestra en la siguiente figura:


![Péndulo Elástico Amortiguado](https://github.com/amalvarezme/SenalesSistemas/blob/master/Otros/circuitoRLC.jpeg?raw=2)
""")
st.write("Calculamos la Ecuación Diferencial")

st.write("La ecuación diferencial que describe la relación entrada/salida en el circuito se obtiene al aplicar las leyes de Kirchhoff. Así, se obtiene la siguiente EDO")

st.latex(r"il(t) = ic(t) + iR(t)")


st.latex(r"CL \frac{d^2 Vo(t)}{dt^2} + \frac{L}{R}\frac{d Vo(t)}{dt} + Vo(t) = Vi(t)")

st.write("Aplicando la trasfomada de laplace tenemos")

st.latex(r"CLs^2Vo(s)+ (LsVo(s)/R) + Vo(s) = Vi(s)")

st.write("""
Tomando transformada de Laplace (condiciones iniciales cero) y usando
$Y(s)$ Y $F(s)$ donde: """)


st.latex(r" H(s) = \frac{Y(s)}{F(s)}")

st.write("Función de transferencia")

st.latex(r"H(s)=\frac{1}{LC s^2 + \frac{Ls}{R} + 1}")

st.write("Comparación con la red eléctrica")

st.write("""
Comparando las EDO del péndulo elástico amortiguado y el circuito RLC, se tiene que ambos constituyen una EDO de segundo orden con coeficientes constantes.

La comparación de la EDO anterior con la del péndulo elástico amortiguado produce equivalencia entre sistemas

|                         | Circuito RLC | Péndulo elástico                          |
|-------------------------|------------------------------------|-------------------------------------------|
| Señal de entrada $x(t)$ | $V_{i}(t) = F_{E}(t) C$  |$F_{E}(t) = 1/C*V_{i}(t)$ |
| Señal de salida $y(t)$  | $V_{o}(t)$                    | $y(t)$                                    |
|                         | $L=m$                              | $m=L$                                     |
|                         | $R=L/Cc$                              | $c=L/RC$                                     |
|                         | $C= 1/k$                    | $k={1}/{C}$                           |
""")


#___________________________________________________________


# ---------------------------

#  (Interfaz interactiva y simulaciones)
# ---------------------------

#  IMPORTACIÓN DE LIBRERÍAS ADICIONALES (si no están ya importadas)
# (las principales ya están importadas arriba: streamlit, numpy, matplotlib, scipy, sympy)
import scipy.signal as signal  # alias consistente
import math

#  CONFIGURACIÓN DE LA PÁGINA (ajuste no intrusivo)
st.set_page_config(page_title="Simulación RLC 2º Orden", layout="wide", initial_sidebar_state="expanded")

# Sidebar (añadimos info adicional sin quitar la del autor)
st.sidebar.write("---")



#titulo de la nueva sección
st.header("Simulación sistema masa–resorte–amortiguador y RLC paralelo (Interactivo)")

st.write("""
Esta herramienta permite simular un sistema masa–resorte–amortiguador y su equivalente
eléctrico RLC en paralelo.
""")

st.markdown(r"""
**Aquí se calculan:**

- Tipo de amortiguamiento: subamortiguado, crítico y sobreamortiguado.
- Frecuencia natural no amortiguada $\omega_n$ y amortiguada $\omega_d$.
- Factor de amortiguamiento $\zeta$.
- Tiempos característicos: pico $t_p$, levantamiento $t_r$ y establecimiento $t_s$.
""")

# TEORÍA (expander)
with st.expander("Modelo mecánico y equivalente RLC paralelo (Resumen)"):
    st.markdown("### Modelo masa–resorte–amortiguador")
    st.latex(r"m\ddot{y}(t) + c\dot{y}(t) + k y(t) = F_E(t)")
    st.markdown("""
    - Entrada mecánica: \\(x(t) = F_E(t)\\) (fuerza externa).
    - Salida: \\(y(t)\\) (desplazamiento de la masa).
    """)
    st.markdown("**Función de transferencia mecánica (condiciones iniciales cero):**")
    st.latex(r"H_m(s) = \frac{Y(s)}{F_E(s)} = \frac{1}{m s^2 + c s + k}")

    st.markdown("### Equivalente RLC en paralelo")
    st.latex(r"H_{RLC}(s) = \frac{V_o(s)}{V_i(s)} = \frac{1}{LC s^2 + \frac{L}{R}s + 1}")
    st.markdown("Comparando denominadores, se obtiene la equivalencia:")
    st.latex(r"m = LC,\quad c = \frac{L}{R},\quad k = 1")
    st.markdown("""
    En este panel se muestran los valores equivalentes \\(R, L, C\\) calculados
    a partir de \\(m\\) y \\(c\\) usando estas relaciones.
    """)

# FUNCIÓN PARA MÉTRICAS DE ESCALÓN
def calcular_metricas_escalon(t_arr, y_arr, valor_final=1.0):
    """
    Calcula:
    - Tiempo de levantamiento (Tr) ~ de 10% a 90%.
    - Máximo sobreimpulso (Mp) [%].
    - Tiempo pico (Tp).
    - Tiempo de establecimiento (Ts) al 2 % (última vez que sale de la banda).
    """
    y = np.array(y_arr)
    t = np.array(t_arr)

    # Tiempo de levantamiento (10% a 90%)
    try:
        idx_10 = np.where(y >= 0.1 * valor_final)[0][0]
        idx_90 = np.where(y >= 0.9 * valor_final)[0][0]
        t_r = t[idx_90] - t[idx_10]
    except:
        t_r = np.nan

    # Máximo sobreimpulso y tiempo pico
    val_max = np.max(y)
    if val_max > valor_final + 1e-12:
        M_p = (val_max - valor_final) / valor_final * 100.0
        idx_pico = np.argmax(y)
        t_p = t[idx_pico]
    else:
        M_p = 0.0
        t_p = np.nan

    # Tiempo de establecimiento (2%): buscamos la última vez que sale de banda ±2%
    banda = 0.02 * valor_final
    t_s = np.nan
    for i in range(len(y)-1, -1, -1):
        if abs(y[i] - valor_final) > banda:
            t_s = t[i]
            break

    return t_r, M_p, t_p, t_s

# SECCIÓN: CONTROLES EN SIDEBAR (interactivo)
st.sidebar.header("Parámetros del sistema (simulación)")

tipo_sistema = st.sidebar.selectbox(
    "Tipo de sistema",
    ["Subamortiguado (0 < ζ < 1)", "Amortiguamiento crítico (ζ = 1)", "Sobreamortiguado (ζ > 1)"]
)

modo_lazo = st.sidebar.radio(
    "Tipo de lazo",
    ["Lazo abierto", "Lazo cerrado"],
    help="En lazo cerrado se usa realimentación unitaria: T(s) = H(s)/(1 + H(s))."
)

# Parámetros mecánicos básicos con sliders (no interfieren con el bloque original)
m_val = st.sidebar.slider("Masa m", 0.1, 10.0, 1.0, 0.1, key="m_sim")
k_val = st.sidebar.slider("Constante del resorte k", 0.1, 20.0, 1.0, 0.1, key="k_sim")

# Selección / rango del factor de amortiguamiento ζ
if tipo_sistema.startswith("Subamortiguado"):
    zeta = st.sidebar.slider("Factor de amortiguamiento ζ (0 < ζ < 1)", 0.05, 0.95, 0.3, 0.05, key="zeta_sub")
elif tipo_sistema.startswith("Sobreamortiguado"):
    zeta = st.sidebar.slider("Factor de amortiguamiento ζ (ζ > 1)", 1.05, 5.0, 2.0, 0.05, key="zeta_sobre")
else:  # Crítico
    zeta = 1.0
    st.sidebar.info("Para amortiguamiento crítico se fija ζ = 1")

# SECCIÓN: CÁLCULOS DEL MODELO (numéricos)
omega_n = np.sqrt(k_val / m_val)
c_val = 2 * zeta * np.sqrt(k_val * m_val)

if 0 < zeta < 1:
    omega_d = omega_n * np.sqrt(1 - zeta**2)
else:
    omega_d = np.nan

if 0 < zeta < 1:
    t_p_analitico = np.pi / omega_d
    t_s_analitico = 3 / (zeta * omega_n)
else:
    t_p_analitico = np.nan
    t_s_analitico = 3 / (zeta * omega_n)

# Equivalente RLC paralelo (usamos L fijo = 1 H para visualizar)
L_eq = 1.0
C_eq = m_val / L_eq if m_val > 0 else np.nan
R_eq = L_eq / c_val if c_val > 0 else np.nan

# Mostrar parámetros como métricas
st.subheader("Parámetros del modelo (simulación)")
col_mec1, col_mec2, col_mec3 = st.columns(3)
col_mec1.metric("m (masa)", f"{m_val:.3g}")
col_mec2.metric("k (resorte)", f"{k_val:.3g}")
col_mec3.metric("c (amortiguador)", f"{c_val:.3g}")

col_dyn1, col_dyn2, col_dyn3 = st.columns(3)
col_dyn1.metric("ωₙ (natural)", f"{omega_n:.4g} rad/s")
col_dyn2.metric("ζ", f"{zeta:.4g}")
col_dyn3.metric("ω_d (amortig.)", f"{omega_d:.4g} rad/s" if 0 < zeta < 1 else "No aplica")

col_tiempo1, col_tiempo2 = st.columns(2)
col_tiempo1.metric("tₚ (analítico)", f"{t_p_analitico:.4f} s" if not np.isnan(t_p_analitico) else "No aplica")
col_tiempo2.metric("tₛ (analítico)", f"{t_s_analitico:.4f} s")

st.markdown("**Equivalente RLC paralelo (m = L·C, c = L/R, k = 1):**")
col_rlc1, col_rlc2, col_rlc3 = st.columns(3)
col_rlc1.metric("R (Ω)", f"{R_eq:.4g}" if not np.isnan(R_eq) else "No definido")
col_rlc2.metric("L (H)", f"{L_eq:.4g}")
col_rlc3.metric("C (F)", f"{C_eq:.4g}" if not np.isnan(C_eq) else "No definido")

# Texto tipo y lazo
texto_tipo = ""
if 0 < zeta < 1:
    texto_tipo = "Sistema subamortiguado: respuesta rápida con oscilaciones y posible sobreimpulso."
elif np.isclose(zeta, 1.0):
    texto_tipo = "Sistema críticamente amortiguado: respuesta más rápida posible sin oscilaciones."
elif zeta > 1:
    texto_tipo = "Sistema sobreamortiguado: respuesta monótona, muy estable pero más lenta."
else:
    texto_tipo = "Valor de ζ fuera del rango típico de análisis clásico."

texto_lazo = "Lazo abierto: muestra la dinámica natural de la planta."
if modo_lazo == "Lazo cerrado":
    texto_lazo = "Lazo cerrado: la realimentación corrige la salida y suele reducir el error y el sobreimpulso."

st.info(f"{texto_tipo}\n\n{texto_lazo}")

# SECCIÓN: FUNCIÓN DE TRANSFERENCIA (NUMÉRICA PARA SIMULACIÓN)
# Forma canónica: H(s) = ωn^2 / (s^2 + 2ζωn s + ωn^2)
num_num = [omega_n**2]
if modo_lazo == "Lazo abierto":
    den_num = [1.0, 2.0*zeta*omega_n, omega_n**2]
else:
    # Para lazo cerrado T(s) = H/(1+H); construimos num y den simbólicamente y luego polinomios numéricos
    # Simplificación práctica: formaremos lazo cerrado num/den mediante algebra simbólica segura
    s_sym = sym.Symbol('s')
    H_sym = (omega_n**2) / (s_sym**2 + 2*zeta*omega_n*s_sym + omega_n**2)
    T_sym = sym.simplify(H_sym / (1 + H_sym))
    # Obtener numerador y denominador simbólicos
    num_c_sym, den_c_sym = sym.fraction(sym.simplify(T_sym))
    # Convertimos a listas numéricas
    num_c_poly = sym.Poly(num_c_sym, s_sym)
    den_c_poly = sym.Poly(den_c_sym, s_sym)
    num_num = [float(coef) for coef in num_c_poly.all_coeffs()]
    den_num = [float(coef) for coef in den_c_poly.all_coeffs()]

# Si estamos en lazo abierto y num_num sigue siendo lista simple, aseguramos floats:
if isinstance(num_num, list):
    num_num = [float(x) for x in num_num]
if isinstance(den_num, list):
    den_num = [float(x) for x in den_num]

# Construcción del sistema numérico con scipy.signal
sistema_num = signal.TransferFunction(num_num, den_num)

# Límite de tiempo para las gráficas
if omega_n > 0 and abs(zeta) > 0.05:
    t_lim = max(5.0, 15.0 / (omega_n * abs(zeta)))
else:
    t_lim = 10.0

st.markdown("---")
st.success("Gráficas generadas con SciPy para el sistema seleccionado")

# SECCIÓN: GRÁFICAS (BODE, polos/ceros, TIEMPO)

# 1. Diagrama de Bode (usando scipy.signal.bode)
st.subheader("1. Diagrama de Bode")
w, mag, phase = signal.bode(sistema_num)
fig_bode, (axm, axp) = plt.subplots(2, 1, figsize=(8, 6), sharex=True)
axm.semilogx(w, mag)
axm.set_ylabel("Magnitud (dB)")
axm.grid(True, which="both", ls=":", alpha=0.5)
axp.semilogx(w, phase)
axp.set_ylabel("Fase (grados)")
axp.set_xlabel("Frecuencia (rad/s)")
axp.grid(True, which="both", ls=":", alpha=0.5)
st.pyplot(fig_bode)
plt.clf()

# 2. Mapa de polos y ceros (usando raíces de polinomios)
st.subheader("2. Mapa de Polos y Ceros")
pz_fig, pz_ax = plt.subplots(figsize=(6,4))
# Polos y ceros
try:
    z_roots = np.roots(num_num) if len(num_num) > 1 else np.array([])
    p_roots = np.roots(den_num)
except Exception:
    z_roots = np.array([])
    p_roots = np.array([])

if z_roots.size:
    pz_ax.scatter(np.real(z_roots), np.imag(z_roots), marker='o', label='Ceros')
if p_roots.size:
    pz_ax.scatter(np.real(p_roots), np.imag(p_roots), marker='x', label='Polos')
pz_ax.axhline(0, color='k', linewidth=0.5)
pz_ax.axvline(0, color='k', linewidth=0.5)
pz_ax.set_xlabel("Parte real")
pz_ax.set_ylabel("Parte imaginaria")
pz_ax.set_title("Mapa polos y ceros")
pz_ax.legend()
pz_ax.grid(True, alpha=0.3)
st.pyplot(pz_fig)
plt.clf()

# 3. Respuestas en el tiempo (impulso, escalón, rampa)
st.subheader("3. Respuestas en el dominio del tiempo")

# Tiempo de simulación
t_sim = np.linspace(0, t_lim, 2000)

# ----------------------------------------
# Respuesta al impulso
# ----------------------------------------
st.markdown("### ➤ Respuesta al impulso")
try:
    t_imp, y_imp = signal.impulse(sistema_num, T=t_sim)
    fig_imp, ax_imp = plt.subplots()
    ax_imp.plot(t_imp, y_imp)
    ax_imp.set_xlabel("Tiempo [s]")
    ax_imp.set_ylabel("Respuesta impulso")
    ax_imp.grid(True, alpha=0.3)
    st.pyplot(fig_imp)
    plt.clf()
except Exception as e:
    st.write("No se pudo calcular la respuesta al impulso:", e)

# ----------------------------------------
# Respuesta al escalón
# ----------------------------------------
st.markdown("### ➤ Respuesta al escalón")
try:
    t_step, y_step = signal.step(sistema_num, T=t_sim)
    fig_step, ax_step = plt.subplots()
    ax_step.plot(t_step, y_step)
    ax_step.set_xlabel("Tiempo [s]")
    ax_step.set_ylabel("Respuesta escalón")
    ax_step.grid(True, alpha=0.3)
    st.pyplot(fig_step)
    plt.clf()
except Exception as e:
    st.write("No se pudo calcular la respuesta al escalón:", e)

# ----------------------------------------
# Respuesta a la rampa
# ----------------------------------------
st.markdown("### ➤ Respuesta a la rampa")
try:
    # r(t) = t (rampa): convolución con respuesta al impulso
    t_imp_r, h_imp = signal.impulse(sistema_num, T=t_sim)
    r = t_sim
    y_ramp = np.convolve(h_imp, r)[:len(t_sim)] * (t_sim[1] - t_sim[0])

    fig_ramp, ax_ramp = plt.subplots()
    ax_ramp.plot(t_sim, y_ramp)
    ax_ramp.set_xlabel("Tiempo [s]")
    ax_ramp.set_ylabel("Respuesta rampa")
    ax_ramp.grid(True, alpha=0.3)
    st.pyplot(fig_ramp)
    plt.clf()
except Exception as e:
    st.write("No se pudo calcular la respuesta a la rampa:", e)


# SECCIÓN: MÉTRICAS A PARTIR DEL ESCALÓN (NUMÉRICO)
st.subheader("4. Métricas de desempeño (a partir de la respuesta al escalón)")

# Si no calculamos y_step arriba por algún error, intentar de nuevo
try:
    _ = y_step
except NameError:
    t_step, y_step = signal.step(sistema_num, T=t_sim)

tr_calc, mp_calc, tp_calc, ts_calc = calcular_metricas_escalon(t_step, y_step)

col_met1, col_met2 = st.columns(2)

txt_tr = f"{tr_calc:.4f} s" if not np.isnan(tr_calc) else "No definido"
txt_mp = f"{mp_calc:.2f} %"
txt_tp = f"{tp_calc:.4f} s" if not np.isnan(tp_calc) else "No aplica"
txt_ts = f"{ts_calc:.4f} s" if not np.isnan(ts_calc) else "No converge"

col_met1.metric("Tiempo de levantamiento Tr (10–90%)", txt_tr)
col_met1.metric("Máximo sobreimpulso Mp", txt_mp)
col_met2.metric("Tiempo pico Tp", txt_tp)
col_met2.metric("Tiempo de establecimiento Ts (2%)", txt_ts)

### Conclusión general sobre las tres respuestas del sistema
st.header("""
Las tres gráficas obtenidas permiten comprender cómo se comporta el sistema dinámico frente a diferentes tipos de señales:

1. **Respuesta al impulso**
   Esta gráfica representa la dinámica más fundamental del sistema, ya que muestra cómo reacciona cuando recibe una entrada instantánea y muy breve.
   Revela si el sistema es oscilatorio, sobreamortiguado o críticamente amortiguado, y permite identificar la rapidez con la que disipa energía.

2. **Respuesta al escalón**
   Modela la reacción del sistema cuando se aplica una entrada constante desde t = 0.
   Es la más útil para evaluar estabilidad, tiempo de establecimiento y si existe sobreimpulso.
   En un sistema sobreamortiguado, la respuesta es lenta y monótona, sin oscilaciones.

3. **Respuesta a la rampa**
   Representa cómo el sistema sigue una entrada creciente linealmente con el tiempo.
   Esta gráfica evidencia la capacidad del sistema para "seguir" señales que cambian continuamente, mostrando el error dinámico y la limitación natural del sistema al responder a variaciones sostenidas.

En conjunto, estas tres respuestas permiten evaluar la dinámica completa del sistema: su rapidez, estabilidad, capacidad de seguimiento y comportamiento frente a distintos tipos de señales que aparecen en aplicaciones reales.
""")

# GRÁFICA COMPARATIVA (sub/crit/sobre)
st.markdown("## GRÁFICA COMPARATIVA")
st.markdown(r"""
Esta gráfica muestra cómo reacciona el sistema ante una entrada escalón variando el factor de amortiguamiento ($\zeta$):
""")

st.subheader("Comparación visual de los tipos de amortiguamiento")

# Parámetros base
m_base = 1.0
k_base = 1.0
wn_base = np.sqrt(k_base / m_base)

# Tres valores de c que representan los casos
c_sub   = 0.6
c_crit  = 2.0
c_sobre = 4.0

z_sub   = c_sub   / (2 * np.sqrt(k_base * m_base))
z_crit  = c_crit  / (2 * np.sqrt(k_base * m_base))
z_sobre = c_sobre / (2 * np.sqrt(k_base * m_base))

num = [wn_base**2]
den_sub   = [1, 2 * z_sub   * wn_base, wn_base**2]
den_crit  = [1, 2 * z_crit  * wn_base, wn_base**2]
den_sobre = [1, 2 * z_sobre * wn_base, wn_base**2]

sys_sub   = signal.TransferFunction(num, den_sub)
sys_crit  = signal.TransferFunction(num, den_crit)
sys_sobre = signal.TransferFunction(num, den_sobre)

t_cmp = np.linspace(0, 12, 2000)
t_sub,   y_sub   = signal.step(sys_sub,   T=t_cmp)
t_crit,  y_crit  = signal.step(sys_crit,  T=t_cmp)
t_sobre, y_sobre = signal.step(sys_sobre, T=t_cmp)

fig_cmp, ax_cmp = plt.subplots(figsize=(7, 4))
ax_cmp.plot(t_sub,   y_sub,   label=fr"Subamortiguado  ($\zeta \approx {z_sub:.2f}$)")
ax_cmp.plot(t_crit,  y_crit,  label=fr"Crítico         ($\zeta = {z_crit:.1f}$)")
ax_cmp.plot(t_sobre, y_sobre, label=fr"Sobreamortiguado ($\zeta \approx {z_sobre:.1f}$)")

ax_cmp.axhline(1.0, color="gray", linestyle="--", linewidth=1)
ax_cmp.set_xlabel("Tiempo [s]")
ax_cmp.set_ylabel("Respuesta al escalón")
ax_cmp.set_title("Comparación de respuestas para distintos factores de amortiguamiento")
ax_cmp.grid(True, alpha=0.3)
ax_cmp.legend(loc="best")

st.pyplot(fig_cmp)
plt.clf()


### Conclusión sobre la gráfica comparativa de amortiguamiento
st.header("""
La gráfica permite visualizar cómo cambia la respuesta de un sistema de segundo orden cuando varía el factor de amortiguamiento (\(\zeta\)):

1. **Subamortiguado (\(\zeta < 1\))**
   El sistema presenta oscilaciones debido a que la energía almacenada en el resorte no se disipa suficientemente rápido.
   La respuesta supera temporalmente el valor final (sobreimpulso) y tarda más tiempo en estabilizarse.

2. **Amortiguamiento crítico (\(\zeta = 1\))**
   Es el punto exacto donde el sistema deja de oscilar pero alcanza el valor final en el menor tiempo posible.
   Representa el comportamiento óptimo cuando se busca rapidez sin sobreimpulso.

3. **Sobreamortiguado (\(\zeta > 1\))**
   El sistema no oscila, pero su respuesta es lenta.
   La disipación de energía es excesiva, causando que el sistema tarde más en llegar al estado estable.

---

###  Interpretación general

La comparación evidencia que **el amortiguamiento controla la velocidad y la estabilidad del sistema**:

- Si hay **demasiado poco amortiguamiento**, el sistema responde rápido pero oscila.
- Si hay **amortiguamiento crítico**, el sistema es rápido y estable.
- Si hay **demasiado amortiguamiento**, el sistema es estable pero lento.

En resumen, la gráfica muestra cómo el valor de \(\zeta\) determina si el sistema prioriza **rapidez**, **estabilidad** o **suavidad** en la respuesta.
""")



```
